```python
import pandas as pd

rows = [
    ["K0", "K0", "A0", "B0"],
    ["K1", "K1", "A1", "B1"],
    ["K2", "K2", "A2", "B2"],
    ["K3", "K3", "A3", "B3"],
]

left = pd.DataFrame(rows, columns=["key1", "key2", "A", "B"])
print(left)
```

```python
import pandas as pd

left = pd.DataFrame({'key1': ['K0', 'K1', 'K2', 'K3'],
                     'key2': ['K0', 'K1', 'K2', 'K3'],
                     'A': ['A0', 'A1', 'A2', 'A3'],
                     'B': ['B0', 'B1', 'B2', 'B3']})
print(left)
right = pd.DataFrame({'key1': ['K0', 'K1', 'K2', 'K3'],
                      'key2': ['K0', 'K1', 'K2', 'K3'],
                      'C': ['C0', 'C1', 'C2', 'C3'],
                      'D': ['D0', 'D1', 'D2', 'D3']})
print(right)

merged = pd.merge(left, right, how='inner', on=['key1', 'key2'])
print(merged)
                     
```
