```python
import pandas as pd

data = [10,11,12]
index = ['a', 'b', 'c']
s = pd.Series(data=data, index=index)
print(s)
print(s['a'])
print(s[0:2])
print(s['a':'b'])
print(s.iloc[0])
print(s.loc['a'])
print(s.iloc[0:2])
print(s.loc['a':'b'])
print(s.iloc[[0, 2]])
print(s.loc[['a', 'c']])
```

```python
mask = [True, False, True]
s[mask]
```

```python
s1 = s.copy()
print(s1)
s1['a'] = 100
print(s1)
print(s)
```

```python
print(s1.replace(to_replace=100, value=300, inplace=True))
print(s1)
```

```python
s1.rename(index={'a': 'A'}, inplace=True)
print(s1)
```

```python
import pandas as pd
s2 =pd.Series([1, 2, 3], index=['a', 'b', 'c'])

data = [100, 110, 120]
index = ['h','k','m']
s3 = pd.Series(data=data, index=index)
result = pd.concat([s2, s3], ignore_index=False)
print(result)
result['j'] = 130
print(result)
```

```python
del result['j']
print(result)
result.drop(labels='a', inplace=True)
print(result)
```

```python
import pandas as pd

left = pd.DataFrame({'key': ['K0', 'K1', 'K2', 'K3'],
                     'A': ['A0', 'A1', 'A2', 'A3'],
                     'B': ['B0', 'B1', 'B2', 'B3']})

right = pd.DataFrame({'key': ['K0', 'K1', 'K2', 'K3'],
                      'C': ['C0', 'C1', 'C2', 'C3'],
                      'D': ['D0', 'D1', 'D2', 'D3']})

merged = pd.merge(left, right, how='inner', on='key')
print(merged)
```
