```python
import numpy as np
import matplotlib
matplotlib.use('nbagg')
%matplotlib inline
import matplotlib.pyplot as plt
```

```python
np.random.seed(0)
x = np.arange(5)
y = np.random.rand(5)
print(x)
print(y)
plt.plot(x, y)
plt.show()
```

```python
np.random.seed(0)
x = np.arange(5)
y = np.random.randint(-5, 5, 5)
fig, ax = plt.subplots(ncols=2)
v_bars = ax[0].bar(x, y, color='c')
h_bars = ax[1].barh(x, y, color='m')
ax[0].axhline(0, color='grey', linewidth=1)
ax[1].axvline(0, color='grey', linewidth=1)
plt.show()
```

```python
fig, ax = plt.subplots()
v_bars = ax.bar(x, y, color='lightblue')
for bar, height in zip(v_bars, y):
    if height < 0:
        bar.set(edgecolor='red', linewidth=2, facecolor='none', color='red')
plt.show()
```

```python
x = np.random.randn(100).cumsum()
print(x)
y = np.linspace(0, 10, 100)

fig, ax = plt.subplots()
ax.fill_between(x, y, color='lightblue')
plt.show()
```

```python
x = np.linspace(0, 10, 200)
y1 = 2*x + 1
y2 = 3*x + 1.2
y_mean = 0.5 *x * np.cos(2*x) + 2.5*x + 1.1
fig, ax = plt.subplots()
ax.fill_between(x, y1, y2, color='lightblue', alpha=0.5)
ax.plot(x, y_mean, color='red', linewidth=2)
plt.show()
```

```python
mean_values = [1,2,3]
variance = [0.2,0.4, 0.5]
bar_label = ['bar1', 'bar2', 'bar3']

x_pos = list(range(len(bar_label)))
max_y = max(zip(mean_values, variance))
plt.ylim([0, (max_y[0] + max_y[1])*1.2])
plt.ylabel('variable y')
plt.xticks(x_pos, bar_label)
plt.bar(x_pos, mean_values, yerr=variance, capsize=5, color='lightblue')
plt.show()
```

```python
x1 = np.array([1, 2, 3])
x2 = np.array([2,3,3])

bar_labels = ['bar1', 'bar2', 'bar3']
fig = plt.figure(figsize=(8, 6))
y_pos = np.arange(len(bar_labels))
plt.barh(y_pos - 0.2, x1, height=0.4, label='x1', color='lightblue')
plt.barh(y_pos + 0.2, x2, height=0.4, label='x2', color='lightgreen', alpha=0.7)
plt.yticks(y_pos, bar_labels)
plt.legend()
plt.show()
```

```python
green_data = np.array([2, 3, 3])
blue_data = np.array([1, 2, 3])
red_data = np.array([3, 2, 1])
labels = ['group 1', 'group 2', 'group 3']
pos = list(range(len(labels)))
width = 0.2
fig = plt.figure(figsize=(8, 6))
plt.bar(pos, green_data, width=width, label='green', color='lightgreen')
plt.bar([p + width for p in pos], blue_data, width=width, label='blue', color='lightblue')
plt.bar([p + 2*width for p in pos], red_data, width=width, label='red', color='lightcoral')
plt.xticks([p + width for p in pos], labels)
plt.legend()
plt.show()
```

```python
data  = range(200, 225, 5)
bar_labels = ['a', 'b', 'c', 'd', 'e']
fig = plt.figure(figsize=(8, 6))
y_pos = np.arange(len(data))
plt.yticks(y_pos, bar_labels, fontsize=12)
bars = plt.barh(y_pos, data, color='lightblue', alpha=0.7)
plt.vlines(min(data), ymin=-0.5, ymax=len(data)-0.5, colors='red', linestyles='dashed')
for b,d in zip(bars, data):
    plt.text(b.get_width() + 1.05, b.get_y() + b.get_height()/2, '{0:.2%}'.format(d/min(data)), va='center')
plt.show()
```

```python
mean_values = range(10,18)
x_pos = range(len(mean_values))

import matplotlib.colors as mcolors
import matplotlib.cm as cm

cmap1= cm.ScalarMappable(mcolors.Normalize(vmin=min(mean_values), vmax=max(mean_values)), cmap=cm.hot)
cmap2= cm.ScalarMappable(mcolors.Normalize(vmin=0, vmax=20), cmap=cm.hot)
plt.subplot(121)
plt.bar(x_pos, mean_values, color=cmap1.to_rgba(mean_values))

plt.subplot(122)
plt.bar(x_pos, mean_values, color=cmap2.to_rgba(mean_values))
plt.show()
```
