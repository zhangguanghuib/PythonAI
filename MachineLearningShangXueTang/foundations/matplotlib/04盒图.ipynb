```python
import numpy as np
import matplotlib
matplotlib.use('nbagg')
%matplotlib inline
import matplotlib.pyplot as plt
```

```python
tang_data= [np.random.normal(0, std, 100) for std in range(1, 4)]
#print(tang_data)
fig = plt.figure(figsize=(8,6))
bplot = plt.boxplot(
    tang_data,
    notch=False,
    sym="s",
    orientation="vertical",
)
plt.xticks([y+1 for y in range(len(tang_data))], ['x1', 'x2', 'x3'])
plt.xlabel("X")
plt.ylabel("Values")
plt.title("Boxplot of Tang Data")
for components in bplot.keys():
    for line in bplot[components]:
        line.set_color('blue')  
plt.show()
```

```python
tang_data= [np.random.normal(0, std, 100) for std in range(1, 4)]
#print(tang_data)
fig = plt.figure(figsize=(8,6))
bplot = plt.boxplot(
    tang_data,
    notch=True,
    sym="s",
    orientation="horizontal",
    patch_artist=True,
)
plt.yticks([y+1 for y in range(len(tang_data))], ['y1', 'y2', 'y3'])
plt.ylabel("Y")
plt.xlabel("Values")
plt.title("Boxplot of Tang Data")
colors = ['blue', 'green', 'red']
for patch, color in zip(bplot["boxes"], colors):
    patch.set_facecolor(color)
plt.show()
```

```python
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(12,6))
tang_data= [np.random.normal(0, std, 100) for std in range(6, 10)]
axes[0].violinplot(tang_data, showmeans=True, showmedians=True)
axes[0].set_title("Violin Plot of Tang Data")
axes[0].set_xlabel("X")
axes[0].set_ylabel("Values")
axes[0].yaxis.grid(True)
axes[0].xaxis.grid(True)

axes[1].boxplot(tang_data)
axes[1].set_title("Box Plot of Tang Data (Horizontal)")
axes[1].set_xlabel("Values")
axes[1].set_ylabel("X")
axes[1].set_yticks([y+1 for y in range(len(tang_data))])
axes[1].set_yticklabels(['y1', 'y2', 'y3', 'y4'])   
axes[1].yaxis.grid(True)
axes[1].xaxis.grid(True)
plt.show()
```
