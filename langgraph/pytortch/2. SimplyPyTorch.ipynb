```python
import torch
from torch import nn

torch.manual_seed(42)

# Generate training data on the CPU.
x = torch.linspace(-1, 1, 200).reshape(-1, 1)
y = x.square()

model = nn.Sequential(
    nn.Linear(1, 32),
    nn.Tanh(),
    nn.Linear(32, 1),
)

loss_fn = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.01)

model.train()
for epoch in range(1000):
    prediction = model(x)
    loss = loss_fn(prediction, y)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    if (epoch + 1) % 200 == 0:
        print(f"Epoch {epoch + 1}: loss = {loss.item():.6f}")

model.eval()
with torch.no_grad():
    test_x = torch.tensor([[0.5]])
    print("Prediction:", model(test_x).item())
    print("Expected:", 0.5 ** 2)
```
