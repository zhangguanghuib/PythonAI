```python
import pandas as pd
import numpy as np
df = pd.DataFrame(np.random.randint(0, 100, (10, 5)), columns=list('ABCDE'))

df.loc[2] = df.loc[6]
df.loc[3] = df.loc[5]
df.loc[8] = df.loc[7]
df.loc[2,'A'] = 100
display(df)
#display(df.duplicated(keep='last', subset=list('BCDE')))
df = df.drop_duplicates(keep='last', subset=list('BCDE'))
display(df)
```

```python
condition = df.duplicated(keep='last', subset=list('BCDE'))
display(condition)
df.loc[condition]
#condition = condition - 1
display(df.loc[(condition-1).astype(bool)])
```

```python
exp_index = df.loc[condition].index
df = df.drop(index=exp_index)
display(df)
```
