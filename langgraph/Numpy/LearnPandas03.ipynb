```python
import pandas as pd
import numpy as np
df = pd.DataFrame(np.random.randn(10000, 4), columns=list('ABCD'))
```

```python
display(df)
```

```python
%matplotlib inline
import matplotlib.pyplot as plt
import seaborn as sns
sns.histplot(df['A'])
plt.show()
```

```python
df.std(axis=0)
```

```python
display(np.abs(df))
```

```python
exception_con = (np.abs(df) > 3* df.std()).any(axis=1)
display(exception_con)
display(df.loc[exception_con])
exp_index = df.loc[exception_con].index
display(exp_index)
display(df.drop(index=exp_index))
```

```python
normal_condition = ~exception_con
display(normal_condition)
display(df.loc[normal_condition])
```
