```python
import pandas as pd
import numpy as np

df = pd.DataFrame(np.random.randint(0, 100, (10, 5)), columns=list('ABCDE'))
display(df)
```

```python
df.rename(mapper=lambda x: str(x)+' Product', axis=1)
```

```python
dic={
    'A': 'A Product',
    'B': 'B Product',
    'C': 'C Product',   
    'D': 'D Product',
    'E': 'E Product'
}
df =df.rename(columns=dic)
display(df)
```
