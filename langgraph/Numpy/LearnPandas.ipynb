```python
index = ['张三', '李四', '王五', '赵六']
columns = ['Python', 'Java', 'C++']
score1 = pd.DataFrame(data=np.random.randint(40, 100, size=(4, 3)), index=index, columns=columns)
display(score1)
score1.loc['张三','Java']= 0
score1.loc['李四'] = score1.loc['李四'] + 10
display(score1)

score2 = pd.DataFrame(data=np.random.randint(40, 100, size=(4, 3)), index=index, columns=columns)
display(score2)
score2 += 10
display(score2)

bool_list = score2['Python'] > score2['Java']
display(bool_list)
display(score2.loc[bool_list])
display(score2.loc[bool_list].index)


score_total = score1 + score2
display(score_total/2)

```

```python
import pandas as pd
import numpy as np

# 行索引：期数 + 销售员
index = pd.MultiIndex.from_product(
    [["第一期", "第二期"], ["alex", "lucy", "tom"]],
    names=["期数", "销售员"]
)

# 按图片中的顺序填写销售数据
data = np.random.randint(0,100,size=(6,3))

df = pd.DataFrame(data, index=index, columns=["A", "B", "C"])
df.columns.name = "产品名称"

display(df)
```
