```python
import time

def call_model():
    print("开始调用模型")
    time.sleep(2)  # 模拟模型调用的延迟
    print("模型调用完成")

def peform_other_tasks():
    for i  in range(5):
        print(f"执行其他任务 {i+1}")
        time.sleep(1)  # 模拟其他任务的延迟

def main():
    start_time = time.time()
    call_model()
    peform_other_tasks()
    end_time = time.time()
    total_time = end_time - start_time
    return f"总耗时: {total_time:.2f} 秒"

if __name__ == "__main__":
    result = main()
    print(result)
```
